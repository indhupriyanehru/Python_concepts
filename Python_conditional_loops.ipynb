```python
#palindrome or not
string = input("enter the string") 
i, j = 0, len(string) - 1  
p = True  

while i < j:
    if string[i] != string[j]:  
        p = False
        break
    i += 1
    j -= 1

if p:
    print("Yes") 
else:
    print("No")
```

```python
#reverse a string
string = input("Enter a string: ")
print(f"the original string is {string}")
rev_string = string[::-1]
print(f"The reversed string is: {rev_string}")
```
